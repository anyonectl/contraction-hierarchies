```python
import pandas as pd

files = {
    "Dijkstra": "results/standard_1000_results.csv",
    "Bidirectional": "results/bidirectional_1000_results.csv",
    "CH": "results_augmented/bidirectional_1000_results.csv"
}

dfs = []
for algo, path in files.items():
    df = pd.read_csv(path)
    df["algorithm"] = algo
    dfs.append(df)

df = pd.concat(dfs, ignore_index=True)

df.head()
```

```python
df["time_ms"] = df["time_ns"] / 1e6 # milliseconds
df["time_μs"] = df["time_ns"] / 1e3 # microseconds
df.head()
```

```python
di = df[df['algorithm'] == "Dijkstra"]
bidi = df[df['algorithm'] == "Bidirectional"]
ch = df[df['algorithm'] == "CH"]

ch.head()
```

```python
print(di.info())
di.describe()
```

```python
print(bidi.info())
bidi.describe()
```

```python
print(ch.info())
ch.describe()
```

```python
time_stats = {
    'Dijkstra': {
        'mean': di['time_μs'].mean(),
        'median': di['time_μs'].median(),
        'min': di['time_μs'].min(),
        'max': di['time_μs'].max()
    },
    'Bidirectional': {
        'mean': bidi['time_μs'].mean(),
        'median': bidi['time_μs'].median(),
        'min': bidi['time_μs'].min(),
        'max': bidi['time_μs'].max()
    },
    'CH': {
        'mean': ch['time_μs'].mean(),
        'median': ch['time_μs'].median(),
        'min': ch['time_μs'].min(),
        'max': ch['time_μs'].max()
    }
}

relaxed_stats = {
    'Dijkstra': {
        'mean': di['relaxed'].mean(),
        'median': di['relaxed'].median(),
        'min': di['relaxed'].min(),
        'max': di['relaxed'].max()
    },
    'Bidirectional': {
        'mean': bidi['relaxed'].mean(),
        'median': bidi['relaxed'].median(),
        'min': bidi['relaxed'].min(),
        'max': bidi['relaxed'].max()
    },
    'CH': {
        'mean': ch['relaxed'].mean(),
        'median': ch['relaxed'].median(),
        'min': ch['relaxed'].min(),
        'max': ch['relaxed'].max()
    }
}
```
